```python
dataset_path = '/content/drive/MyDrive/Google_Earth_Engine_labelled_dataset - 2'
save_drive_path = '/content/drive/MyDrive/YoloV8_Tree_Enumeration_Results'
```

```python
from ultralytics import YOLO

model = YOLO('yolov8n.yaml')

results = model.train(
    data=f'{dataset_path}/data.yaml',
    epochs=100,
    imgsz=640,
    batch=16,
    lr0=0.01,
    save_period=1,
    project=f"{save_drive_path}/runs",
    name="custom_model",
    save=True
)
```

```python
import pandas as pd
import matplotlib.pyplot as plt

results_csv_path = '/content/drive/MyDrive/YoloV8_Tree_Enumeration_Results/runs/custom_model/results.csv'

results_df = pd.read_csv(results_csv_path)

print(results_df.head())

plt.figure(figsize=(12, 6))

plt.subplot(1, 2, 1)
plt.plot(results_df['epoch'], results_df['train/box_loss'], label='Train Box Loss')
plt.plot(results_df['epoch'], results_df['train/cls_loss'], label='Train Class Loss')
plt.title('Training Loss over Epochs')
plt.xlabel('Epochs')
plt.ylabel('Loss')
plt.legend()

plt.subplot(1, 2, 2)
plt.plot(results_df['epoch'], results_df['metrics/mAP50(B)'], label='mAP@0.5')
plt.plot(results_df['epoch'], results_df['metrics/mAP50-95(B)'], label='mAP@0.5:0.95')
plt.title('Mean Average Precision over Epochs')
plt.xlabel('Epochs')
plt.ylabel('mAP')
plt.legend()

plt.tight_layout()
plt.show()
```

```python
import gradio as gr
from ultralytics import YOLO
import cv2
from PIL import Image
import numpy as np

model = YOLO('/content/drive/MyDrive/YoloV8_Tree_Enumeration_Results/runs/custom_model/weights/best.pt')

def detect_trees(input_image):
    image = np.array(input_image)

    results = model(image)

    annotated_image = results[0].plot()

    return Image.fromarray(annotated_image)

with gr.Blocks(css="""
    body {
        background: linear-gradient(135deg, #2c3e50, #3498db);  /* Gradient background */
        color: #ffffff;  /* White text */
        font-family: 'Arial', sans-serif;
    }
    .gr-button {
        background-color: #e74c3c;  /* Red button */
        color: white;
        border: none;
        border-radius: 12px;
        padding: 10px 20px;
        font-size: 16px;
        font-weight: bold;
        cursor: pointer;
        transition: background-color 0.3s ease;
    }
    .gr-button:hover {
        background-color: #c0392b;  /* Darker red on hover */
    }
    .gr-image {
        border-radius: 10px;
        border: 3px solid #ffffff;  /* White border around image */
    }
    .gr-box {
        border-radius: 12px;
        padding: 20px;
        background-color: rgba(255, 255, 255, 0.1);  /* Semi-transparent box for inputs/outputs */
    }
    .gr-description, .gr-title {
        color: #ffffff;  /* Ensure title and description are white */
        text-align: center;
    }
    footer {display: none;}  /* Hide the flag button */
""") as demo:

    gr.Markdown(
        """
        <h1 style="font-size: 2.5em; margin-bottom: 20px;">Tree Detection with YOLOv8</h1>
        <p style="font-size: 1.2em;">Upload an image, and the model will detect trees and show bounding boxes.</p>
        """)

    with gr.Row():
        with gr.Column():
            image_input = gr.Image(type="pil", label="Upload an Image", elem_id="input_image")
        with gr.Column():
            image_output = gr.Image(type="pil", label="Annotated Image", elem_id="output_image")

    detect_button = gr.Button("Detect Trees", elem_id="detect_button")

    detect_button.click(detect_trees, inputs=image_input, outputs=image_output)

demo.launch()
```

```python
import pandas as pd

results_csv_path = '/content/drive/MyDrive/YoloV8_Tree_Enumeration_Results/runs/custom_model/results.csv'

try:
    results_df = pd.read_csv(results_csv_path)
    print("Results CSV loaded successfully.")
except FileNotFoundError:
    print(f"Error: The file {results_csv_path} does not exist.")
    exit()

required_columns = ['metrics/precision(B)', 'metrics/recall(B)', 'metrics/mAP50(B)', 'metrics/mAP50-95(B)']
if not all(col in results_df.columns for col in required_columns):
    print("Error: Some required columns are missing from the results CSV.")
    exit()

precision = results_df['metrics/precision(B)'].iloc[-1]
recall = results_df['metrics/recall(B)'].iloc[-1]
mAP50 = results_df['metrics/mAP50(B)'].iloc[-1]
mAP50_95 = results_df['metrics/mAP50-95(B)'].iloc[-1]

if precision + recall != 0:
    accuracy_approx = (2 * precision * recall) / (precision + recall)
else:
    accuracy_approx = 0

print(f"Precision: {precision:.4f}")
print(f"Recall: {recall:.4f}")
print(f"mAP@0.5: {mAP50:.4f}")
print(f"mAP@0.5:0.95: {mAP50_95:.4f}")
print(f" Accuracy: {accuracy_approx:.4f}")
```

```python
import pandas as pd

results_csv_path = '/content/drive/MyDrive/YoloV8_Tree_Enumeration_Results/runs/custom_model/results.csv'

results_df = pd.read_csv(results_csv_path)

precision = results_df['metrics/precision(B)'].iloc[-1]
recall = results_df['metrics/recall(B)'].iloc[-1]
mAP50 = results_df['metrics/mAP50(B)'].iloc[-1]
mAP50_95 = results_df['metrics/mAP50-95(B)'].iloc[-1]
inference_speed = '2.1 ms (per image)'
post_process_speed = '1.1 ms (per image)'

if precision + recall != 0:
    f1_score = (2 * precision * recall) / (precision + recall)
else:
    f1_score = 0

evaluation_metrics = {
    'Parameters': ['Model', 'Epochs', 'Image Size', 'Accuracy', 'IoU Threshold',
                   'Save Period', 'Precision', 'Recall', 'mAP@50', 'mAP@50-95', 'Inference Speed',
                   'Post Processing Speed'],
    'Values': ['YOLOv8', len(results_df), 640, round(f1_score, 4), 0.5, 1, round(precision, 4),
               round(recall, 4), round(mAP50, 4), round(mAP50_95, 4), inference_speed, post_process_speed]
}

df_metrics = pd.DataFrame(evaluation_metrics)

print(df_metrics)

from IPython.display import display
display(df_metrics)
```
